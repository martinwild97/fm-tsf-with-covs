**Distance calculation of each time series pair (Libra benchmark):**

The distance between two time series is calculated using the $L_1$ norm of their describing time series characteristics. This approach is adopted from the zoomed ranking method applied to time series.

The function for calculating the distance between two time series, $Y_i$ and $Y_j$, is defined as:

$d(Y_i, Y_j) = \sum_{m=1}^{q} \left| \frac{c_{Y_i,m} - c_{Y_j,m}}{\max_{k=1,...,s}(c_{Y_k,m}) - \min_{k=1,...,s}(c_{Y_k,m})} \right|$

Where:
* $Y_1, ..., Y_s$ are all time series from all considered data sets, with $s$ being a positive integer.
* $Y_i$ and $Y_j$ originate from the same data set.
* $c_{Y_i,1}, ..., c_{Y_i,q}$ are the describing time series characteristics of time series $Y_i$, with $q$ being a positive integer (specifically, 25 characteristics are used).

The distance value ranges from 0 to infinity, where 0 indicates that the time series are identical in terms of their describing characteristics, and a higher distance signifies greater heterogeneity between the time series.

**Selection of timeseries**

- Takes the timeseries that has the biggest overall distance first
- Adds the next time series that has the biggest distance to already selected timeseries
- Same steps in a loop up to a given number of timeseries
- Focus on integration of all datasets: checks if all datasets are present. If not selcts the next timeseries only out of missing datasets (min ts per dataset: 5)
